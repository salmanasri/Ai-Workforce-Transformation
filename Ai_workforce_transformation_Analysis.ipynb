<h2 style="color:#B54708;">⚠️ 2.8 Outlier Detection</h2>

<p>
Outliers are unusually high or low values.
</p>
<p>
The <b>Interquartile Range (IQR)</b> method is used to detect potential
outliers in the numerical features.
</p>

```python
df[numerical_columns].describe().T
```

```python
# Calculate IQR(Interquartile Range) for numerical columns

Q1 = df[numerical_columns].quantile(0.25)
Q3 = df[numerical_columns].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outlier_count = ((df[numerical_columns] < lower_limit) |
                 (df[numerical_columns] > upper_limit)).sum()

print(outlier_count.sort_values(ascending=False))
```

<p style="margin:0;">
<b>❓</b>How can industries be grouped based on their AI transformation, workforce preparation, and business outcomes?
</p>

</div>

<p>
Industries are grouped into descriptive profiles based on three dimensions:
<strong>AI Transformation</strong> (AI adoption and automation),
<strong>Workforce Preparation</strong> (AI training, reskilling and workforce resilience),
and <strong>Productivity Outcome</strong>.
</p>

<p>
Employment change and salary growth are kept separate from the classification so that the industry profiles focus on
<strong>AI transformation and workforce preparedness</strong>. These employment and salary measures can then be examined
separately as business outcomes.
    <p><b>"These profiles are descriptive groups created from standardized industry-level indicators and should not be interpreted as official industry classifications or performance rankings."</b></P>
</p>


```python
#Create the three dimensions

# AI Transformation
industry_scaled["ai_transformation"] =(
    industry_scaled["ai_adoption_rate"] +
    industry_scaled["automation_level"]
) / 2

# Workforce Preparation
industry_scaled["workforce_preparation"] =(
    industry_scaled["employee_ai_training_hours"] +
    industry_scaled["reskilling_rate"] +
    industry_scaled["workforce_resilience_score"]
) / 3

# Productivity Outcome
industry_scaled["productivity_outcome"] = (
    industry_scaled["productivity_gain_percent"]
)
```

```python
# Assign the profiles

def classify_industry(row):

    ai = row["ai_transformation"]
    workforce = row["workforce_preparation"]
    productivity = row["productivity_outcome"]
                                                        
    if ai > 0 and workforce > 0 and productivity > 0:            # High AI adoption, strong workforce readiness, and positive productivity improvement.
        return "Advanced AI Transformation"                     # A company uses AI widely, trains/reskills employees, and achieves higher productivity.

   
    elif ai > 0 and workforce > 0:          # High AI adoption and strong workforce readiness, but productivity improvement is not necessarily positive.
        return "Strong Transformation"      # A company has adopted AI and prepared employees well, but productivity improvement has not yet appeared.

   
    elif ai > 0 and productivity <= 0:           # High AI adoption with low or no productivity improvement, suggesting a stronger focus on automation.
        return "Automation-Focused Transformation" #A company is using AI mainly for automation, but productivity has not improved.

   
    elif ai <= 0 and workforce <= 0 and productivity <= 0:          # Low AI adoption, low workforce readiness, and low or no productivity improvement.
        return "Lower AI Transformation"                       #A company has limited AI use, little workforce preparation, and little productivity improvement.

   
    else:
        return "Moderate / Emerging Transformation"               # A mixed transformation pattern that does not clearly fit into the other categories.
                                                        #A company may have some AI adoption but either workforce readiness or productivity is still developing

industry_scaled["profile"] = industry_scaled.apply(
    classify_industry,
    axis=1
)
```
